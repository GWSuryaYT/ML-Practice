# Unsupervised Learning:

Before in supervised learning we passed the label of the data but now we wont. before we passed {xi, yi} where xi is the feature vector with d-dimensonal data and yi is class/object label. After we will only pass {xi} and group the data.

Eg:

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

# Create a Toy Dataset

We'll generate a dataset with 300 points and 3 centers from clustering.

```python
# Generate synthetic dataset
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)

#plot the dataset 
plt.scatter(X[:,0], X[:,1], s=50,c='Gray') #here x[:,0] is taking out all the 0th indexed value outof last list try to see them below
plt.title("Generated Data")
plt.show()
```

# Initialize KMeans

We choose K=3 clusters (since we generated 3 centers). We'll also set random_state for reproducibility

```python
#initialize Kmeans:
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=0)
```

# Fit KMeans to the data:

The algorithm will assign each point to a cluster and compute the cluster centers.

```python
#fit the model
kmeans.fit(X)

#get the cluster centers:
centers = kmeans.cluster_centers_

#get the labels for each points:
labels = kmeans.labels_
```

# Visualize the clusters:

We will plot the colored by cluster label and mark the clusters centers

```python
#plot clusters
plt.scatter(X[:,0],X[:,1], c=labels, s=50, cmap='viridis')
plt.scatter(centers[:,0], centers[:,1], c='red', s=200, marker='X')
plt.title('KMeans Clustering Result')
plt.show()
```

# Use elbow method to find optimal k

We calculate the sum of squared distances (inertia) for diff values of k

```python
inertia = []
K=range(1,10)
for k in K:
    kmeans = KMeans(n_clusters=k,init='k-means++',random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.plot(K, inertia, 'bx-')
plt.xlabel('Number of clusters K')
plt.ylabel('Inertia (sum of squared distances)')
plt.title('Elbow method to find optimial K')
plt.show()
```
