#Trying to Simulate SGD by adding Noices :>


\(L(w)=w^{4}-2w^{2}+0.5w\)

```python
import numpy as np
import matplotlib.pyplot as plt

def l(w):
    return w**4 - 2*w**2 + 0.5*w

#Define the stochastic gradient (with noise)
def grad_l(w,noice_level=0.5):

    noice_data = np.random.normal(0, noice_level)

    return (4*w**3 - 4*w + 0.5)+noice_data

tol = 1e-6
k = 0
max_iter = 50
alpha = 0.05
point = 2 #our starting point

history_of_x = [point]
history_of_y = [l(point)]

while np.linalg.norm(grad_l(point))>tol and max_iter > k:

    #direction=
    direction = -grad_l(point, 0.6)

    #new_pos
    new_pos = point + (direction * alpha)

    point = new_pos

    history_of_x.append(point)
    history_of_y.append(l(point))

    k = k + 1

#graph plot:
X= np.linspace(-2,2,100)
Y= l(X)
plt.plot(X,Y, color= 'Blue')

#plot the gradient path:
plt.plot(history_of_x, history_of_y, color='Red', marker='o')
plt.show()
```
