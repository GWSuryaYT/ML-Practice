1D function:

f(x) = x**2

```python
#how to plot a function in python:

import matplotlib.pyplot as plt
import numpy as np

# 1. Define the x-axis range (Start, Stop, Number of points)
x = np.linspace(-10, 10, 500)

# 2. Define your function
y = x**2 

# 3. Create the plot
plt.plot(x, y, label="y = x^2", color="blue")

# 4. Customize (Always add labels!)
plt.title("Plot of a Quadratic Function")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.legend()

# 5. Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return x**2

def grad_f(x):
    return 2*x

data_history_points = [-5]

tol = 0
max_iter=500

#learning rate:
alpha = 0.2 #(step_length)

k = 0
point = -5

while np.linalg.norm(grad_f(point)) > tol and max_iter > k:

    #direction:
    direction = -grad_f(point)

    #new postion
    new_point = point + (alpha * direction)

    #update
    point = new_point

    #update
    k = k + 1

    data_history_points.append(point)

    print(f'Current Height: {f(point)}, epoch: {k}, Current Point: {point}')

#lets plot smooth underlying 1D function curve

x_curve = np.linspace(-6, 6, 400)
y_curve = f(x_curve)
plt.plot(x_curve, y_curve, label="f(x) = x^2", color="gray", linestyle="--")

#map the history to function heights:
x_history = np.array(data_history_points)
y_history = f(x_history)
plt.plot(x_history, y_history, color= 'red', marker = 'o', label='GD steps')
plt.show()

```

In 2D functions when your point already contains (x,y) those times you use the function to get Z's value that will tell you the height or altitude.

```python
# 2D: f(x,y) = x**2 + y**2

def f(x,y):
    return x**2 + y**2

def grad_f(x,y):
    return np.array([2*x, 2*y])

point = np.array([4,5])
tol = 1e-6
k=0
alpha = 0.1
max_iter = 200

#collection of history of points:
history_of_x = [point[0]]
history_of_y = [point[1]]
history_of_z = [f(point[0], point[1])]

while np.linalg.norm(grad_f(point[0], point[1])) > tol and max_iter > k:

    #direction:
    d = -grad_f(point[0], point[1])

    #new position:
    new_pos = point + (alpha * d)

    #update:
    point = new_pos

    #update:
    k = k + 1

    #add to history:
    history_of_x.append(new_pos[0])
    history_of_y.append(new_pos[1])
    history_of_z.append(f(new_pos[0], new_pos[1]))

    print(f'Current height {f(point[0], point[1])}, Current Point {point}, Current Epoch {k}')



# 3. Plotting in 3D
fig = plt.figure(figsize=(10, 14))
ax = fig.add_subplot(projection='3d')

# Generate the smooth 3D surface underneath
x_surf = np.linspace(-5, 5, 100)
y_surf = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x_surf, y_surf)
Z = f(X, Y)
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6) # The 3D landscape

ax.plot(history_of_x, history_of_y, history_of_z, color='red', marker='.', 
    linewidth=2, label='2D GD Path')

ax.set_title("Gradient Descent Trajectory on a 3D Surface")
ax.set_xlabel("X position")
ax.set_ylabel("Y position")
ax.set_zlabel("Z (Function Output)")
ax.legend() 
plt.show()
```

```python
#2d
import numpy as np
import matplotlib.pyplot as plt

def f (x,y):
    return (x+2*y-7)**2 + (2*x +y -5)**2

def gradf(x,y):
    df_x = 10*x + 8*y - 34
    df_y = 8*x + 10*y -38

    return np.array([df_x, df_y])

point = [8,7]
k=0
alpha = 0.1
tol=1e-6
max_iter = 1000
history_of_x= [point[0]]
history_of_y= [point[1]]
history_of_z= [f(point[0], point[1])] #we need this bcz its the height


while np.linalg.norm(gradf(point[0], point[1])) > tol and max_iter>k:

    #direction:
    d = -gradf(point[0], point[1])

    #new position:
    new_pos = point + (alpha * d)

    #update:
    point = new_pos

    #update:
    k = k + 1

    history_of_x.append(point[0])
    history_of_y.append(point[1])
    history_of_z.append(f(point[0], point[1]))


    print(f'Current Height: {f(point[0], point[1])}, Current Positiong {point}, Epoch: {k}')


# 3. Plotting in 3D
fig = plt.figure(figsize=(10, 14))
ax = fig.add_subplot(projection='3d')

# Generate the smooth 3D surface underneath
x_surf = np.linspace(-10, 10, 100)
y_surf = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x_surf, y_surf)
Z = f(X, Y)
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6) # The 3D landscape

ax.plot(history_of_x, history_of_y, history_of_z, color='red', marker='.', 
    linewidth=2, label='2D GD Path')

ax.set_title("Gradient Descent Trajectory on a 3D Surface")
ax.set_xlabel("X position")
ax.set_ylabel("Y position")
ax.set_zlabel("Z (Function Output)")
ax.legend() 
plt.show()

fig2, ax2 = plt.subplots(figsize=(8, 6))
contours = ax2.contour(X, Y, Z, levels=30, cmap='viridis')
ax2.clabel(contours, inline=True, fontsize=8)
ax2.plot(history_of_x, history_of_y, color='red', marker='o', markersize=4, linewidth=2, label='2D GD Path')
ax2.scatter(1, 3, color='blue', marker='*', s=150, zorder=5, label='Global Minimum (1,3)')
ax2.set_title("2D Contour Map Trajectory")
ax2.set_xlabel("X position")
ax2.set_ylabel("Y position")
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

plt.show()
```
