### 1. Why a Negative Directional Derivative Guarantees Going Downhill

Imagine you are standing on a foggy mountain surface, and $f(x)$ represents your altitude.

* The **gradient vector** $\nabla f(x_k)$ is a compass pointing in the direction of the **steepest uphill slope**.
* The **directional derivative** measures your rate of altitude change if you walk in direction $d_k$. Mathematically, it is calculated using the dot product:

$$\text{Directional Derivative} = \nabla f(x_k) \cdot d_k$$



If you choose your direction $d_k$ to be directly opposite to the gradient, $d_k = -\nabla f(x_k)$ (steepest descent):


$$\nabla f(x_k) \cdot (-\nabla f(x_k)) = -\Vert{}\nabla f(x_k)\Vert{}^2$$

Since the square of any magnitude is always positive, adding that negative sign guarantees the result is **always negative**.

A negative rate of altitude change literally means: **"As you move forward in direction $d_k$, your height decreases."** That is why setting $d_k$ such that $\nabla f(x_k) \cdot d_k < 0$ mathematically guarantees you are taking a downhill path.

### 2. The Meaning of $f(x_k + \alpha_k d_k) < f(x_k)$

Break down the components inside the brackets:

* **$x_k$**: Your current $(x, y)$ coordinate vector on the map.
* **$d_k$**: The vector direction you decided to walk in.
* **$\alpha_k$**: The scalar stride length (step size).
* **$x_k + \alpha_k d_k$**: Your **prospective new coordinate** after taking that step.

Now wrap them in $f(\dots)$:

* **$f(x_k)$**: Your current altitude (height).
* **$f(x_k + \alpha_k d_k)$**: The altitude at your prospective new coordinate.

The inequality $f(x_k + \alpha_k d_k) < f(x_k)$ simply states: **"My height at the new position must be strictly lower than my height at the old position."**

If your step size $\alpha_k$ is too large, you might leap straight over the valley bottom and land higher up on the opposite hill. Checking this condition ensures you don't overshoot.

```python
import numpy as np
import matplotlib.pyplot as plt
```

Let's use a classic test function: a simple 3D bowl defined by the equation \(f(x, y) = x^2 + y^2\). 

```python
def f(x):
    """
    This function accepts a point as i said before we need a starting point lets say its X it have to be a vector like (1,2) smt so use np.array([x1,x2]) smt like that
    """

    #x is a numpy array made by eg: np.array([3,9])
    #we will extract that:
    x_cords = x[0]
    y_cords = x[1]

    #now we will use the target equation: f(x,y) = x^2 + y^2
    result = (x_cords ** 2) + (y_cords ** 2)

    return result

def gradf (x):
    """
    Calculates the gradient vector (the direction of steepest uphill).
    Returns a NumPy array containing [df/dx, df/dy].
    """
    df_dx = 2 * x[0]
    df_dy = 2 * x[1]

    return np.array([df_dx, df_dy])

#our starting position x or current position:
current_position = np.array([3.0, 4.0])

#our height Z= 
height = f(current_position)
print(f'Current Height: {height}')

# 3. find the uphill direction: gradient:
uphill_copass = gradf(current_position)
print(f'Uphill Gradient Vector: {uphill_copass}') # [2 * 3 , 2* 4] = [6 , 8]

# 4. find the downhill direction (steepest descept):
#remember from blueprint: d = -gradient
downhill_direction = -uphill_copass
print(f'Downhill direction Vector: {downhill_direction}')
```

Now that I have a downhill direction, the next logical phase is to actually take a step and move your position on the map:

```python
#step:
alpha = 0.1

#2. calculate your new cords on the map: same formula:
#new_position = xK+1 = oldpos(xK) +step_size (aK) x direction(dK):
new_position = current_position + (alpha * downhill_direction)

#3.check your new height:
new_height = f(new_position)

print(f"Old Position: {current_position} | Old Height: {f(current_position)}")
print(f"New Position: {new_position} | New Height: {new_height}")
```

see our new position is getting low now its time to pack everything to a loop until we reach bottom:

That is a fantastic question! If the gradient is a direction, why do we care about its length (the magnitude/steepness), and why do we use that length to stop the loop?The short answer is: The length of the gradient tells us how close we are to the bottom. As you get closer to the lowest point of a smooth valley, the ground flattens out, and the length of the gradient shrinks toward zero.

```python
def f(x):
    return x[0]**2 + x[1]**2

def grad_f(x):
    return np.array([2* x[0], 2* x[1]])



x = np.array([3.0,4.0])

#---------------------Loop---------------------------
#1. step:
alpha = 0.2

# step counter :
k = 0

#setting a safety:
max_iter = 1000

#Tolerance: tol = 1e-6 (defines when the slope is close enough to 0):
tol = 1e-6

print(f"Starting at position {x} with height {f(x)}\n")

while np.linalg.norm(grad_f(x)) > tol and k < max_iter:
    
    #get the downhill direction:
    d = -grad_f(x)

    #update the point by taking a step:
    x_new = x + alpha * d

    #apply the new position:
    x = x_new

    #update the iter:
    k +=1

    # Print the first few 10 steps and final steps to see progress
    if k <= 10 or np.linalg.norm(grad_f(x)) <= tol:
        print(f"Step {k}: Coordinates = {x}, Height = {f(x)}, Slope Steepness = {np.linalg.norm(grad_f(x))}")

print(f"\nOptimization Finished!")
print(f"Total Steps Taken (k): {k}")
print(f"Final Minimum Coordinates: {x}")
print(f"Final Minimum Height: {f(x)}")
```
